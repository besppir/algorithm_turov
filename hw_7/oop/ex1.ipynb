```python
class Vector:
    def __init__(self, x, y=None, z=None):
        if isinstance(x, str) and y is None and z is None:
            cleaned = x.strip().strip('{}')
            coords = [float(num.strip()) for num in cleaned.split(',')]
            assert len(coords) == 3, "Строка должна содержать ровно 3 координаты"
            self.x, self.y, self.z = coords
        else:
            assert isinstance(x, (int, float)), "Координата x должна быть числом"
            assert isinstance(y, (int, float)), "Координата y должна быть числом"
            assert isinstance(z, (int, float)), "Координата z должна быть числом"
            self.x = x
            self.y = y
            self.z = z

    def __abs__(self):
        return (self.x**2 + self.y**2 + self.z**2)**0.5

    def __add__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return Vector(self.x + other.x, self.y + other.y, self.z + other.z)

    def __sub__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return Vector(self.x - other.x, self.y - other.y, self.z - other.z)

    def __mul__(self, other):
        if isinstance(other, Vector):
            return self.x * other.x + self.y * other.y + self.z * other.z
        elif isinstance(other, (int, float)):
            return Vector(self.x * other, self.y * other, self.z * other)
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)
    
    def cross(self, other):
        return Vector(
            self.y * other.z - self.z * other.y,
            self.z * other.x - self.x * other.z,
            self.x * other.y - self.y * other.x
        )
    def __repr__(self):
        return f"Vector({self.x}, {self.y}, {self.z})"

points = [
    Vector("{1, 2, 3}"),
    Vector("{4, 5, 6}"),
    Vector("{7, 8, 9}"),
    Vector("{-1, 0, 2}"),
    Vector("{3, -2, 5}")
]

total = Vector(0, 0, 0)
for p in points:
    total = total + p

n = len(points)
center = total * (1 / n)

print(f"Координаты центра масс: x = {center.x:.1f}, y = {center.y:.1f}, z = {center.z:.1f}")

max_area = 0
max_A = None
max_B = None
max_C = None

for i in range(n):
    for j in range(i + 1, n):
        for k in range(j + 1, n):
            A = points[i]
            B = points[j]
            C = points[k]

            AB = B - A
            AC = C - A
            area = 0.5 * abs(AB.cross(AC))
            
            if area > max_area:
                max_area = area
                max_A = A
                max_B = B
                max_C = C

print("Максимальная площадь:", max_area)
print("Точки:")
print("A =", max_A)
print("B =", max_B)
print("C =", max_C)

```
