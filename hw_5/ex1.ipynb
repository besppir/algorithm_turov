```python
s = input()
parts = s.split('student_')[1:]

nums = [p[:3] for p in parts]
scores = [int(p[3:]) for p in parts]
table = dict(zip(nums, scores))
m = max(scores)
keys = [n for n, sc in zip(nums, scores) if sc == m] # вдруг номер повторится, поэтому лучше чем начальное k for k, v in table.items() if v == m
print('-'.join(keys))
```
