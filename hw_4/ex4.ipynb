```python
# import matplotlib.pyplot as plt
# import numpy as np
# import pandas as pd

# f = pd.read_csv('iris_data.csv')

# SepalLengthCm = list(f['SepalLengthCm'])
# SepalWidthCm = list(f['SepalWidthCm'])
# PetalLengthCm = list(f['PetalLengthCm'])
# PetalWidthCm = list(f['PetalWidthCm'])


# fig = plt.figure(figsize = (16,9))
# ax1 = fig.add_subplot(231)
# ax2 = fig.add_subplot(232)
# ax3 = fig.add_subplot(233)
# ax4 = fig.add_subplot(234)
# ax5 = fig.add_subplot(235)
# ax6 = fig.add_subplot(236)

# ax1.plot(SepalWidthCm, SepalLengthCm, 'r', label='SepalLengthCm from SepalWidthCm')
# k, b = np.polyfit(SepalWidthCm, SepalLengthCm, 1)
# x_line = np.linspace(min(SepalWidthCm), max(SepalWidthCm), 100)
# ax1.plot(x_line, k * x_line + b, 'b-', label=f'МНК: y={k:.2f}x+{b:.2f}')
# ax1.set_xlabel('SepalWidthCm')
# ax1.set_ylabel('SepalLengthCm')
# ax1.legend() дальше не хахотелось кааждой прямой свой мнк


# ax2.plot(PetalLengthCm, SepalLengthCm, 'r', label='SepalLengthCm from PetalLengthCm')
# ax2.set_xlabel('PetalLengthCm')
# ax2.set_ylabel('SepalLengthCm')
# ax2.legend()

# ax3.plot(PetalWidthCm, SepalLengthCm, 'r', label='SepalLengthCm from PetalWidthCm')
# ax3.set_xlabel('PetalWidthCm')
# ax3.set_ylabel('SepalLengthCm')
# ax3.legend()

# ax4.plot(PetalLengthCm, SepalWidthCm, 'r', label='SepalWidthCm from PetalLengthCm')
# ax4.set_xlabel('PetalLengthCm')
# ax4.set_ylabel('SepalWidthCm')
# ax4.legend()

# ax5.plot(PetalWidthCm, SepalWidthCm, 'r', label='SepalWidthCm from PetalWidthCm')
# ax5.set_xlabel('PetalWidthCm')
# ax5.set_ylabel('SepalWidthCm')
# ax5.legend()

# ax6.plot(PetalWidthCm, PetalLengthCm, 'r', label='PetalLengthCm from PetalWidthCm')
# ax6.set_xlabel('PetalWidthCm')
# ax6.set_ylabel('PetalLengthCm')
# ax6.legend()

# plt.show()

import matplotlib.pyplot as plt #точки графика отсортированы чтобы не было каши как в случае 1 + все было бы компактнее
import numpy as np
import pandas as pd

f = pd.read_csv('iris_data.csv')

SepalLengthCm = list(f['SepalLengthCm'])
SepalWidthCm = list(f['SepalWidthCm'])
PetalLengthCm = list(f['PetalLengthCm'])
PetalWidthCm = list(f['PetalWidthCm'])

pairs = [
    (SepalWidthCm, SepalLengthCm, 'SepalWidthCm', 'SepalLengthCm'),
    (PetalLengthCm, SepalLengthCm, 'PetalLengthCm', 'SepalLengthCm'),
    (PetalWidthCm, SepalLengthCm, 'PetalWidthCm', 'SepalLengthCm'),
    (PetalLengthCm, SepalWidthCm, 'PetalLengthCm', 'SepalWidthCm'),
    (PetalWidthCm, SepalWidthCm, 'PetalWidthCm', 'SepalWidthCm'),
    (PetalWidthCm, PetalLengthCm, 'PetalWidthCm', 'PetalLengthCm'),
]

fig = plt.figure(figsize=(16, 9))

for idx, (x, y, xlabel, ylabel) in enumerate(pairs, start=1):

    pairs_sorted = sorted(zip(x, y))
    x_sorted = [p[0] for p in pairs_sorted]
    y_sorted = [p[1] for p in pairs_sorted]

    ax = fig.add_subplot(2, 3, idx)
    ax.plot(x_sorted, y_sorted, 'r', label=f'{ylabel} from {xlabel}')

    k, b = np.polyfit(x_sorted, y_sorted, 1)
    x_line = np.linspace(min(x_sorted), max(x_sorted), 100)
    ax.plot(x_line, k * x_line + b, 'b-', label=f'МНК: y={k:.2f}x+{b:.2f}')

    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.legend(fontsize=8)

    print(f'{ylabel} от {xlabel}: k={k:.3f}, b={b:.3f}')

plt.tight_layout()
plt.show()
```
