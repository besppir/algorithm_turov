```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

fig = plt.figure(figsize = (16,9))
ax1 = fig.add_subplot(221)
ax2 = fig.add_subplot(222)
ax3 = fig.add_subplot(223)

sm = np.random.normal(0, 10, 100)
md = np.random.normal(0, 10, 1000)
lg = np.random.normal(0, 10, 10000)

ax1.hist(sm, 10)
ax2.hist(md, 100)
ax3.hist(lg, 1000) 

plt.show()
```
