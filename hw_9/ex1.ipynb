```python

def build_tree(a):
    n = len(a)
    tree = {}
    
    for i in range(n):
        tree[n + i] = a[i]
        
    for i in range(n - 1, 0, -1):
        tree[i] = max(tree[2 * i], tree[2 * i + 1])
        
    return tree


def max_tree(tree, n, l, r):
    _max = float('-inf')
    l += n - 1
    r += n
    
    while l < r:
        if l % 2 == 1:
            _max = max(_max, tree[l])
            l += 1
            
        if r % 2 == 1:
            r -= 1
            _max = max(_max, tree[r])
            
        l //= 2
        r //= 2
        
    return _max


N = int(input())
a = list(map(int, input().split()))
K = int(input())

tree = build_tree(a)

answers = []

for i in range(K):
    l, r = map(int, input().split())
    answers.append(max_tree(tree, N, l, r))

print(*answers)
```
