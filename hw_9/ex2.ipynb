```python

from math import gcd

def build_tree(a):
    n = len(a)
    tree = [0] * (2 * n)

    for i in range(n):
        tree[n + i] = a[i]

    for i in range(n - 1, 0, -1):
        tree[i] = gcd(tree[2 * i], tree[2 * i + 1])

    return tree


def gcd_tree(tree, n, l, r):
    result = 0

    l += n - 1
    r += n

    while l < r:
        if l % 2 == 1:
            result = gcd(result, tree[l])
            l += 1

        if r % 2 == 1:
            r -= 1
            result = gcd(result, tree[r])

        l //= 2
        r //= 2

    return result


N = int(input())
a = list(map(int, input().split()))
K = int(input())

tree = build_tree(a)

answers = []

for i in range(K):
    l, r = map(int, input().split())
    answers.append(gcd_tree(tree, N, l, r))

print(*answers)
```
