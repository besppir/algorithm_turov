```python
class Node:
    def __init__(self, key=None, left=None, right=None):
        self.key = key
        self.left = left
        self.right = right


def isMirrorTrees(x, y):
    if x is None and y is None:
        return True
    if x is None or y is None:
        return False
    return (x.key == y.key
            and isMirrorTrees(x.left, y.right) 
            and isMirrorTrees(x.right, y.left))


def isSymmetric(root):
    if root is None:
        return True
    return isMirrorTrees(root.left, root.right)
```
