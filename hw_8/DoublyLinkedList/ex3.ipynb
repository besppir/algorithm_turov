```python
class Node:
    def __init__(self, data):
        self.item = data
        self.nref = None


class LinkedList:
    def __init__(self):
        self.start_node = None

    def insert_at(self, pos, data):
        if pos < 0:
            print("Позиция не может быть отрицательной")
            return
        new_node = Node(data)
        if pos == 0:
            new_node.nref = self.start_node
            self.start_node = new_node
            return
        n = self.start_node
        for i in range(pos - 1):
            if n is None:
                break
            n = n.nref
        if n is None:
            print("Позиция вне списка")
            return
        new_node.nref = n.nref
        n.nref = new_node

    def delete_by_value(self, x):
        if self.start_node is None:
            print("Список пуст")
            return
        if self.start_node.item == x:
            self.start_node = self.start_node.nref
            return
        n = self.start_node
        while n.nref is not None and n.nref.item != x:
            n = n.nref
        if n.nref is None:
            print("Элемент не найден")
        else:
            n.nref = n.nref.nref

    def print_list(self):
        if self.start_node is None:
            print("Список пуст")
            return
        n = self.start_node
        while n is not None:
            print(n.item, end=" ")
            n = n.nref

    def differences(self):
        if self.start_node is None:
            return []
        n = self.start_node
        while n.nref is not None:
            n = n.nref
        last = n.item
        result = []
        n = self.start_node
        while n is not None:
            result.append(n.item - last)
            n = n.nref
        return result


lst = LinkedList()
for i, v in enumerate([5, 8, 2, 11, 7]):
    lst.insert_at(i, v)

print(*lst.differences()) 
lst.delete_by_value(11)
lst.print_list()
print(lst.differences())

lst.insert_at(1, 9)
lst.print_list()
print(lst.differences())
```
